```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error
plt.rcParams['font.sans-serif'] = ['Simhei']  #显示中文
plt.rcParams['axes.unicode_minus'] = False    #显示负号
plt.rcParams["figure.figsize"] = (12, 8)

# 读取数据
data = pd.read_csv('boston.csv', sep=',')

# 基本信息
print(data.head(10))
print(data.info())
print(data.describe())
print(data.isnull().sum())

# 可视化
data.hist(bins=30, figsize=(20,15))
plt.suptitle('特征分布直方图', fontsize=20)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

plt.figure(figsize=(20,15))
for i, column in enumerate(data.columns, 1):
    plt.subplot(4, 4, i)
    sns.kdeplot(data[column], shade=True)
    plt.title(f'{column} 分布')
plt.tight_layout()
plt.show()

# 计算相关系数矩阵
corr_matrix = data.corr()
plt.figure(figsize=(12,10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm')
plt.title('特征相关性热力图')
plt.show()
# 不同特征与因变量“MEDV”（房价中值）间的相关性
plt.figure(figsize=(20, 15))
for i, feature in enumerate(data.columns[:-1], 1):
    plt.subplot(4, 4, i)
    sns.scatterplot(x=data[feature], y=data['MEDV'])
    sns.regplot(x=data[feature], y=data['MEDV'], scatter=False, color='red')
    plt.xlabel(feature)
    plt.ylabel('MEDV')
plt.tight_layout()
plt.show()

# 数据预处理
data_filled = data.fillna(data.mean())
features = data_filled.columns[:-1]
# 使用Z-Score方法进行数据预处理
data_zscore = data_filled.copy()
data_zscore[features] = stats.zscore(data_filled[features])

# 划分训练集和测试集并训练模型
X = data_zscore.drop('MEDV', axis=1).values
y = data_zscore['MEDV'].values
random_seeds = [42, 7, 21, 100, 2024]
results = []

def linear_regression_normal_equation(X, y):
    X_b = np.hstack([np.ones((X.shape[0], 1)), X])
    theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
    return theta

def linear_regression_gradient_descent(X, y, lr=0.01, epochs=1000):
    X_b = np.hstack([np.ones((X.shape[0], 1)), X])
    theta = np.zeros(X_b.shape[1])
    for epoch in range(epochs):
        predictions = X_b.dot(theta)
        errors = predictions - y
        gradients = (2/X_b.shape[0]) * X_b.T.dot(errors)
        theta -= lr * gradients
        if epoch % 100 == 0:
            mse = np.mean(errors**2)
            print(f'迭代 {epoch} 次: MSE={mse:.4f}')
    return theta

def evaluate_model(theta, X, y):
    X_b = np.hstack([np.ones((X.shape[0], 1)), X])
    predictions = X_b.dot(theta)
    mse = mean_squared_error(y, predictions)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y, predictions)
    return mse, rmse, mae

def train_and_evaluate(X_train, y_train, X_test, y_test):
    theta_normal = linear_regression_normal_equation(X_train, y_train)
    theta_gd = linear_regression_gradient_descent(X_train, y_train)
    mse_normal, rmse_normal, mae_normal = evaluate_model(theta_normal, X_test, y_test)
    mse_gd, rmse_gd, mae_gd = evaluate_model(theta_gd, X_test, y_test)
    return {
        'Normal Equation': {'MSE': mse_normal, 'RMSE': rmse_normal, 'MAE': mae_normal},
        'Gradient Descent': {'MSE': mse_gd, 'RMSE': rmse_gd, 'MAE': mae_gd}
    }

for seed in random_seeds:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, shuffle=True, random_state=seed)
    metrics = train_and_evaluate(X_train, y_train, X_test, y_test)
    results.append({
        '随机种子': seed,
        '正规方程 MSE': metrics['Normal Equation']['MSE'],
        '正规方程 RMSE': metrics['Normal Equation']['RMSE'],
        '正规方程 MAE': metrics['Normal Equation']['MAE'],
        '梯度下降 MSE': metrics['Gradient Descent']['MSE'],
        '梯度下降 RMSE': metrics['Gradient Descent']['RMSE'],
        '梯度下降 MAE': metrics['Gradient Descent']['MAE'],
    })

results_df = pd.DataFrame(results)
print(results_df)

```
