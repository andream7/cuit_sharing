```python
from pulp import LpMaximize, LpProblem, LpVariable

# 创建线性规划模型
model = LpProblem("利润最大化", LpMaximize)

# 定义变量
x_A = LpVariable("x_A", lowBound=0, cat='Continuous')  # 产品A的生产件数
x_B = LpVariable("x_B", lowBound=0, cat='Continuous')  # 产品B的生产件数
x_C = LpVariable("x_C", lowBound=0, cat='Continuous')  # 产品C的生产件数

# 目标函数
model += 4 * x_A + 2 * x_B + 3 * x_C, "总利润"

# 约束条件
model += 7 * x_A + 3 * x_B + 6 * x_C <= 150, "劳动力约束"
model += 4 * x_A + 4 * x_B + 5 * x_C <= 200, "材料约束"

# 求解模型
status = model.solve()

# 输出结果
print("产品A的生产件数:", x_A.varValue)
print("产品B的生产件数:", x_B.varValue)
print("产品C的生产件数:", x_C.varValue)
print("总收益:", model.objective.value())
```

```python
from pulp import LpMinimize, LpProblem, LpVariable, LpInteger

# 创建线性规划模型
model = LpProblem("Minimize_Nurse_Count", LpMinimize)

# 定义变量，每个变量表示从对应班次开始的护士人数
x1 = LpVariable("x1", lowBound=0, cat=LpInteger)  
x2 = LpVariable("x2", lowBound=0, cat=LpInteger) 
x3 = LpVariable("x3", lowBound=0, cat=LpInteger)  
x4 = LpVariable("x4", lowBound=0, cat=LpInteger) 
x5 = LpVariable("x5", lowBound=0, cat=LpInteger) 
x6 = LpVariable("x6", lowBound=0, cat=LpInteger)  

# 目标函数：最小化总的护士人数
model += x1 + x2 + x3 + x4 + x5 + x6

# 约束条件：确保每个班次的最小护士需求得到满足
model += x1 + x4 + x5 + x6 >= 60
model += x2 + x5 + x6 + x1 >= 70
model += x3 + x6 + x1 + x2 >= 60
model += x4 + x1 + x2 + x3 >= 50
model += x5 + x2 + x3 + x4 >= 20
model += x6 + x3 + x4 + x5 >= 30

# 求解模型
model.solve()

# 输出结果
print("从班次1开始的护士人数:", x1.varValue)
print("从班次2开始的护士人数:", x2.varValue)
print("从班次3开始的护士人数:", x3.varValue)
print("从班次4开始的护士人数:", x4.varValue)
print("从班次5开始的护士人数:", x5.varValue)
print("从班次6开始的护士人吕:", x6.varValue)
print("总护士人数:", model.objective.value())
```

```python
from pulp import LpMinimize, LpProblem, LpVariable, LpInteger

# 创建线性规划模型
model = LpProblem("最小化总成本", LpMinimize)

# 定义变量
x1 = LpVariable("x1", lowBound=0, cat=LpInteger)  # 一级检验员人数
x2 = LpVariable("x2", lowBound=0, cat=LpInteger)  # 二级检验员人数

# 目标函数
model += 40 * x1 + 36 * x2, "总成本"

# 约束条件
model += 200 * x1 + 120 * x2 >= 1800, "最低产量"

# 求解模型
status = model.solve()

# 输出结果
x1_value = x1.varValue
x2_value = x2.varValue
total_cost = model.objective.value()

print("一级检验员人数:", x1_value)
print("二级检验员人数:", x2_value)
print("总费用:", total_cost)
```
