```python
def backtrack(t):
    global f1, f2, f, x, M, bestf, bestx, n
    if t >= n:
        bestx = x[:]
        bestf = f
    else:
        for j in range(t, n):
            f1 += M[x[j]][1]
            k = t - 1
            f2[t] = max(f2[k], f1) + M[x[j]][2]
            f += f2[t]
            if f < bestf:
                x[t], x[j] = x[j], x[t]
                backtrack(t + 1)
                x[t], x[j] = x[j], x[t]
            f1 -= M[x[j]][1]
            f -= f2[t]

import sys

M = [[0, 0, 0], [0, 2, 1], [0, 3, 1], [0, 2, 3]]

f = 0
f1 = 0
n = len(M)

f2 = [i for i in range(n)]
x = [i for i in range(n)]

bestf = sys.maxsize
bestx = None

backtrack(1)

print("最优解为:", bestx)
print("最优值为:", bestf)
```
